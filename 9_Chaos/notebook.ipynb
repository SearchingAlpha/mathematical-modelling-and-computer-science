```python
s = 10.
r = 30.
b = 3.0
Dt = 0.01

def initialize():
    global x, y, z, xresult, yresult, zresult, t, timesteps
    x = y = z = 1.0
    xresult = [x]
    yresult = [y]
    zresult = [z]

    t = 0.0
    timesteps = []

def observe():
    global x, y, z, xresult, yresult, zresult
    xresult.append(x)
    yresult.append(y)
    zresult.append(z)
    timesteps.append(t)

def update():
    global x, y, z, xresult, yresult, zresult, t, timesteps
    nextx = x + Dt * (s * (y - x))
    nexty = y + Dt * (x * (r - z) - y)
    nextz = z + Dt * (x * y - b * z)
    x, y, z = nextx, nexty, nextz
    t += Dt

initialize()
while t < 30.: 
    update()
    observe()

fig = figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot(xresult, yresult, zresult, lw=1)

show()
```

```python
rvalues = arange(0, 40, 5)

s = 10.
b = 3.0
Dt = 0.01

def initialize():
    global x, y, z, xresult, yresult, zresult, t, timesteps
    x = y = z = 1.0
    xresult = [x]
    yresult = [y]
    zresult = [z]

    t = 0.0
    timesteps = []

def observe():
    global x, y, z, xresult, yresult, zresult
    xresult.append(x)
    yresult.append(y)
    zresult.append(z)
    timesteps.append(t)

def update():
    global x, y, z, xresult, yresult, zresult, t, timesteps
    nextx = x + Dt * (s * (y - x))
    nexty = y + Dt * (x * (r - z) - y)
    nextz = z + Dt * (x * y - b * z)
    x, y, z = nextx, nexty, nextz
    t += Dt

def lorenz():
    initialize()
    while t < 30.:
        update()
        observe()
    return xresult, yresult, zresult

for r_val in rvalues:
    global r
    r = r_val
    xresult, yresult, zresult = lorenz()

    fig = figure()
    ax = fig.add_subplot(111, projection='3d')
    ax.plot(xresult, yresult, zresult, lw=1)
    ax.set_title(f"Lorenz Attractor (r = {r_val})")

show()
```
