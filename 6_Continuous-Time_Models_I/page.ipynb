### Code 6.1
logistics-growth-continous model

```python
r = 0.2
K = 1.
Dt = 0.01

def initialize():
    global x, result, t, timesteps
    x = 0.1
    result = [x]
    t = 0.
    timesteps = [t]

def observe():
    global x, result, t, timesteps
    result.append(x)
    timesteps.append(t)

def update():
    global x, result, t, timesteps
    x = x + r * x * (1-x / K) * Dt
    t += Dt

initialize()
while t < 50.:
    observe()
    update()

plot(timesteps, result)
show()
```

### Exercise 6.5
Simulate the following continous time Lotka-Volterra (predator-prey) model

```python
Dt = 0.0001
def initialize():
    global x, xresult, y, yresult, t, timesteps
    x = 0.1
    y = 0.1
    xresult = [x]
    yresult = [y]
    t = 0.
    timesteps = [t]

def observe():
    global x, xresult, y, yresult, t, timesteps
    xresult.append(x)
    yresult.append(y)
    timesteps.append(t)

def update():
    global x, xresult, y, yresult, t, timesteps
    nextx = x + (x-x*y)*Dt
    nexty = y + (x*y - y)*Dt
    x = nextx
    y = nexty
    t += Dt

initialize()
while t < 50.:
    observe()
    update()
plot(timesteps, xresult, label='x')
plot(timesteps, yresult, label='y')
legend()
show()
```

### Exercise 6.8
Imagine two chemical species, S and E interacting in a test tube.
S + E -> 2E
Develop a continuous-time mathematical model that describes the temporal changes of the concentration of S and E and simulate its behavior

```python
Dt = 0.001
k = 1.
def initialize():
    global s, e, sresult, eresult, t, timesteps
    s = 0.99
    e = 0.01
    sresult = [s]
    eresult = [e]
    t = 0.
    timesteps = [t]

def observe():
    global s, e, sresult, eresult, t, timesteps
    sresult.append(s)
    eresult.append(e)
    timesteps.append(t)

def update():
    global s, e, sresult, eresult, t, timesteps
    nexts = s - (k*s*e) * Dt
    nexte = e + (k*s*e) * Dt
    s = nexts
    e = nexte
    t += Dt

initialize()
while t < 20.:
    observe()
    update()
plot(timesteps, sresult, label='s')
plot(timesteps, eresult, label='e')
legend()
show()
```
