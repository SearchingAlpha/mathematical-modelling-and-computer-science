```python
# Sawtooth function
# Generate x values from 0 to 2π
T = 5.0

# Generate x values over a few periods
x = np.linspace(0, 6 * T, 500)

# Compute f(x) = x mod T
y = np.mod(x, T)

# Plot
plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('x mod T')
plt.title(f'Plot of f(x) = x mod T (T = {T})')
plt.grid(True)
```

```python
# Sinusoidal function

x = np.linspace(0, 4*np.pi, 200)

# Plot cos(x)**a for a in 1..5
for a in range(1, 5):
    y = np.cos(x) ** a
    plt.plot(x, y, label=f'a = {a}')

# Labeling and styling
plt.xlabel('x')
plt.ylabel('cos(x)^a')
plt.title('Plot of cos(x)^a for a from 1 to 5')
plt.legend()
plt.grid(True)

# Display the plot
plt.show()
```
